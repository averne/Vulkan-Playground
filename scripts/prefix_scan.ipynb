```python
import numpy as np
```

```python
dat  = np.random.randint(16, size=2**24,     dtype=np.uint8)
flag = np.random.randint(2,  size=dat.shape, dtype=np.uint8)
# flag = np.zeros_like(dat)
# flag[:len(dat) >> 1] = 1

dat .tofile("in.bin")
flag.tofile("flag.bin")
```

```python
def inclusive_sum(arr):
    a = arr.copy()
    for i in 2**np.arange(len(a).bit_length()):
        a[i:] += a[:-i]
    return a

def exclusive_sum(arr):
    return inclusive_sum(arr) - arr

def inclusive_sum_with_flag(arr, flag):
    a, f = arr.copy(), flag.copy()
    for i in 2**np.arange(len(a).bit_length()):
        a[i:] += np.where(f[i:] == 0, a[:-i], 0)
        f[i:] |= f[:-i]
    return a

def exclusive_sum_with_flag(arr, flag):
    return inclusive_sum_with_flag(arr, flag) - arr
```

```python
dat = dat.astype(np.uint32)
s = inclusive_sum_with_flag(dat, flag)
s.tofile("out-ref.bin")

np.dstack((dat, flag, s))
```

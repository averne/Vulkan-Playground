```python
import numpy as np, matplotlib.pyplot as plt
```

```python
dat = np.random.randint(16, size=2**24, dtype=np.uint8)
dat.tofile("in.bin")
```

```python
s = np.cumsum(dat, dtype=np.uint32)
s.tofile("out-ref.bin")
```

```python
np.cumsum(dat.reshape((-1, 256)), axis=1, dtype=np.uint32).tofile("out-ref2.bin")
```

```python
np.array_equal(s, np.fromfile("out.bin", dtype=np.uint32))
```
